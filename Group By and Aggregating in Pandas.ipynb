# Group by and Aggregating

#### Group by is going to group together the values in the column and display them all in the same row and it allows in aggregation

```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\vidhi\OneDrive\Desktop\coding\Pandas\data\csv.files\Flavors.csv")
```

```python
df
```

```python
group_by_frame = df.groupby('Base Flavor')
```

```python
group_by_frame
```

```python
group_by_frame.mean(numeric_only = True)
```

```python
# Combining the above groupby and aggregate funtion in one line.
df.groupby('Base Flavor').mean(numeric_only = True)
```

```python
df.groupby('Base Flavor').count()
```

```python
df.groupby('Base Flavor').min()
```

```python
df.groupby('Base Flavor').max()
# (In the flavor values are coming according to alphabetical order).
```

```python
df.groupby('Base Flavor').sum(numeric_only = True)
```

```python
# Using a separate aggregate function for all these aggregate calculations(sum,count,mean,etc).
df.groupby('Base Flavor').agg({'Flavor Rating':['max','min','mean','sum','count']})
# What we need to pass in our aggregate function is actually a dictionary.
```

```python
df.groupby('Base Flavor').agg({'Flavor Rating':['mean','min','max','sum','count'],'Texture Rating':['mean','min','max','sum','count']})
```

```python
df
```

```python
# We can group on multiple columns
df.groupby(['Base Flavor','Liked']).mean(numeric_only = True)
```

```python
df.groupby(['Base Flavor','Liked']).agg({'Flavor Rating':['mean','max','min','sum','count']})
```

```python
# All these aggregate function can be given by using describe() function which gives a high level overview
df.groupby('Base Flavor').describe()
```
