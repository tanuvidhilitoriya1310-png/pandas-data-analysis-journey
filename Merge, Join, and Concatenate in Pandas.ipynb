# Merge, Join, and Concatenate

```python
import pandas as pd
```

```python
# we will be using 2 different dataframes(left and right) which would be used for joining etc.
df1 = pd.read_csv(r"C:\Users\vidhi\OneDrive\Desktop\coding\Pandas\data\csv.files\LOTR.csv")
df1
```

```python
df2 = pd.read_csv(r"C:\Users\vidhi\OneDrive\Desktop\coding\Pandas\data\csv.files\LOTR 2.csv")
df2
```

## Merge

```python
df1.merge(df2)
# it is default(it gives inner join by default).
# df1 becomes left dataframe and df2 becomes right dataframe automatically.
```

```python
df1.merge(df2, how = 'inner') 
# 'how' tells us how do we merge the dataframes
```

```python
df1.merge(df2, how = 'inner', on = 'FellowshipID')
# 'on' tells us on which column would be choosen to compare for merge
# In the current case we took only one column from which we would join df
# So other same common columns are included by providing them suffixes
```

```python
df1.merge(df2, how = 'inner', on = ['FellowshipID','FirstName'])
```

```python
df1.merge(df2, how ='outer')
# the outer provides every field of dataframe whether they match or not.
# It automattically write NaN(Not a Number) which does not have any field.
# Overlapping data will not be duplicated.
```

```python
df1.merge(df2, how = 'left')
# left join pulls everything from the left df and pulls the remaining overlapping data which we can pull from the right df.
```

```python
df1.merge(df2, how = 'right')
# We are pulling all the data from right df and merging only the overlapping data from the left df.
```

```python
df1.merge(df2, how = 'cross')
# it takes each value from the left df and compare each value from the right df
```

## Join

```python
# It does not have so many default functionality as Merge.
df1.join(df2, on = 'FellowshipID', lsuffix = '_left', rsuffix = '_right')
# It does not really make much sense.
```

```python
df1.join(df2, on = 'FellowshipID' , how = 'outer', lsuffix = '_Left', rsuffix = '_Right')
```

```python
# We are using index to join the df properly and to make some sense.
df3 = df1.set_index('FellowshipID').join(df2.set_index('FellowshipID'), on = 'FellowshipID', lsuffix = '_left', rsuffix = '_right')
df3
# We can use 'how' to define the type of join.
```

## Concatenate

#### Concatenate means putting one df on top of another df ,instead of putting one df  on side of another.

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], join = 'inner')
```

```python
pd.concat([df1,df2], join = 'outer', axis = 1)
```
