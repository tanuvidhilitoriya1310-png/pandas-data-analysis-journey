# Data Cleaning in Pandas

```python
import pandas as pd
```

```python
df= pd.read_excel(r"C:\Users\vidhi\OneDrive\Desktop\coding\Pandas\data\excel.files\Customer Call List.xlsx")
df
```

```python
# 1. Most case drop the duplicates
df = df.drop_duplicates()
df
```

```python
# 2.Remove any columns we do not need
df.drop(columns = 'Not_Useful_Column',inplace = True)
# To save the changes in our df ,eithere assign the df to it just like in first step or use inplace 
```

```python
df.columns = df.columns.str.strip()
df
```

```python
# 3. Now we will check column one by one for any errors we need to clean


# df['Last_Name'] = df['Last_Name'].str.lstrip('...')
# df['Last_Name'] = df['Last_Name'].str.lstrip('/')
# df['Last_Name'] = df['Last_Name'].str.rstrip('_')


df['Last_Name']=df['Last_Name'].str.strip('123./_')
df



# strip is just like trim in sql it either strip from left side (lstrip) or right side(rstrip) and we can specify the value.
# by default it removeswhitespaces.
# Replace is another good option for this case
```

```python
df['Phone_Number']=df['Phone_Number'].str.replace('[^a-zA-Z0-9]','',regex = True)
df
# We are using regex here
# '^'should return anything except inside it
```

```python
# In lambda we should format on string
df.loc[:,'Phone_Number']=df['Phone_Number'].apply(lambda x: str(x))
df
# ':' modifies all rows in 'Phone_Number' column.
```

```python
# Now we will format it, We can format it using lambda or for loop
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + "-" + x[3:6] + "-" + x[6:10])
df
```

```python
# Lets get rid of Nan-- and na--
df['Phone_Number'] = df['Phone_Number'].replace('nan--','')

df['Phone_Number'] = df['Phone_Number'].replace('Na--','')

df
```

```python
df[['Street_Address','State','Zip_code']] = df['Address'].str.split(',',n = 2, expand = True)

df
# 'n' is used to define how many separator we need to sperate our string
# .str.split() provides list by default so we use ,expand = True to separate it
```

```python
df['Paying Customer'] = df['Paying Customer'].str.replace('Yes','Y')
df['Paying Customer'] = df['Paying Customer'].str.replace('No','N')

df
```

```python
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('Yes','Y')
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('No','N')

df
```

```python
df = df.replace('N/a','')
df = df.fillna('')
df
# TO fill any missing value we use .fillna() function.
```

## Remove those people who do not have phone number or we cannot contact them.

```python
for x in df.index:
    if df.loc[x,'Do_Not_Contact'] == 'Y' :
        df.drop(x, inplace = True)
        
df
```

```python
for x in df.index:
    if df.loc[x,'Phone_Number'] == '':
        df.drop(x, inplace = True)

df

# Another way to drop null values
# df.dropna(subset = 'Phone_Number', inplace = True)
# df
```

```python
df = df.drop(columns = 'Address')
```

```python
# We are dropping the Zip_code column because it does not contain any value, that's why it is useless to us.
df.drop(columns='Zip_code',inplace=True)
```

```python
# TO provide new index 
df = df.reset_index(drop = True)

df
# drop = True , drops the original index.
```
