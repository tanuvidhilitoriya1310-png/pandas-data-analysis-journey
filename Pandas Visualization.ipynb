# Pandas Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df =  pd.read_csv(r"C:\Users\vidhi\OneDrive\Desktop\coding\Pandas\data\csv.files\Ice Cream Ratings.csv")
df=df.set_index('Date')
df
```

## Line plot

```python
# The way visualize something in pandas is that we use something called plot
df.plot()
# By default it gives line plot
```

```python
df.plot(kind = 'line')
# the 'kind' defines what plot we want to make.
```

```python
df.plot(kind = 'line',subplots = True)
# subplots is used when we want to see the plot of every column separately.
```

```python
# We can Change the visualization using matplotlib
print(plt.style.available)
# It shows all the style which are available to change the visualization
plt.style.use('fast')
```

```python
df.plot(kind = 'line', title = 'Ice Cream Ratings', xlabel = 'Daily Rating', ylabel = 'Scores')
# title gives the name of the plot and x and y label provides their names
```

## Bar Plot

```python
df.plot(kind = 'bar')
```

```python
# We can stack the columns onto each other using stacked
df.plot(kind = 'bar', stacked = True)
```

```python
# We could specify a particular column in this stacked bar chart.
# It removes the legends because it shows us multiple values but we have taken only one value, so it is removed.
df['Flavor Rating'].plot(kind = 'bar',stacked = True)
# It automatically gives us a vertical bar chart as default.
```

```python
# We could turn the bar chart or stacked bar chart horizontal.
df.plot.barh(stacked = True)
```

## Scatter Plot

```python
# We need to specify x-axis and y-axis for scatter plot to work.
df.plot.scatter(x = 'Flavor Rating', y = 'Overall Rating')
```

```python
# we can change the size of dots using 's' and we can change the color of the dots using'c'.
df.plot.scatter(x='Flavor Rating', y='Overall Rating', s = 400, c ='Pink')
```

## Histogram

#### Histogram describe the distribution of variables (Frequency Distribution)
#### Bins define the ranges or interval in which data are divided, bins default value is 10.

```python
df.plot.hist(bins = 10)
```

# Box Plot

#### Minimum = Smallest value (excluding outliers)
#### Q1 (1st Quartile) = 25% of data lies below this
#### Median (Q2) = Middle value (50%)
#### Q3 (3rd Quartile) = 75% of data lies below this
#### Maximum = Largest value (excluding outliers)

```python
'''
Box = From Q1 to Q3 (called IQR: Interquartile Range)
Line inside box = Median
Whiskers = Extend to min and max
Dots outside = Outliers
'''
df.boxplot()
```

## Area Plot

```python
df.plot.area(figsize = (10,5))
# we could use figsize to  increase the size of the plot
```

## Pie Chart

```python
df.plot.pie(y = 'Flavor Rating',figsize = (10,10))
# We need to specify what column we are working on
```
