```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\vidhi\OneDrive\Desktop\coding\Pandas\data\csv.files\world_population.csv")
df
```

```python
pd.set_option('display.float_format', lambda x: '%.2f' % x)

# pandas is used to control how floating-point numbers are displayed.
```

```python
df.info()
```

```python
# Check for null or missing value in the dataset
df.isnull().sum()
```

```python
# Check for Unique values
df.nunique()
# In the 'World Population Percentage column we were given 70 unique values because most of the values are very small so they were count as one
```

```python
df.sort_values(by='World Population Percentage',ascending = False).head()
```

```python
# We are performing correlation on numeric columns to know their relationship.
df.corr(numeric_only=True)
```

```python
# Let us visualize this correlation using heatmaps
plt.figure(figsize=(12,10))
sns.heatmap(df.corr(numeric_only=True),annot=True)
plt.show()
```

```python
# Use case: Are their certain continents that are growing faster and  in which way?
df.groupby('Continent').mean(numeric_only=True).sort_values(by='2022 Population',ascending=False)
```

```python
df[df['Continent'].str.contains('Oceania')]
```

```python
#lets visualize our use case
```

```python
df2=df.groupby('Continent')[['1970 Population',
       '1980 Population', '1990 Population', '2000 Population',
       '2010 Population', '2015 Population', '2020 Population',
       '2022 Population']].mean(numeric_only=True).sort_values(by='2022 Population',ascending=False)
df2

#For a shortcut you could have used [df.columns[5:13]],but it is not really great in our case.
```

```python
df.columns
```

```python
df3=df2.transpose() 
df3
```

```python
df3.plot(figsize=(10,5))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
# After examining this plot we could se that at 2000 Asia has a huge increase in population 
# Oceania's population rate remained same throught the years
# There is a drastic difference between Asia and oher continent in population.
```

```python
# We will be using box plot to find outliers and dig into the data for more details.
df.boxplot(figsize=(12,7))
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
# In normal cases we would look in the outlier but in our population data it is not needed.
```
